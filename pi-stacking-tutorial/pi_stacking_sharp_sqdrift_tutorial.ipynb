## Tutorial: modelling dispersion on a DNA-inspired $\pi$-stacking system with SHARP-qDRIFT

In this tutorial, we estimate the energy of the same $\pi$-stacked system as in the [BEAST-VQE $\pi$-stacking tutorial](pi_stacking_beast_tutorial.ipynb), but now with SHARP-qDRIFT, a sample-based quantum diagonalization algorithm. We focus on two points of the scan, the equilibrium geometry and the dissociated limit, and use them to estimate the binding energy of the stacked dimer. The quantum chemistry problem is built with projective embedding, and the quantum circuits are simulated with Kvantify's memory-restricted simulator.

<div align="center"><img src="images/pi_stacking_image.png" alt="Thymine dimer extracted from the DNA atomistic model PDB 208D" width="450"></div>

**Why is this interesting?** The stacking of neighbouring nucleobases is one of the interactions that hold the DNA double helix together. It is a dispersion interaction: it is weak (about 1 kcal/mol for the dimer studied here) and it originates entirely from electron correlation. Mean-field methods, and many common density functionals such as B3LYP, miss it altogether, which makes $\pi$-stacking a classic stress test for correlated methods.

SHARP-qDRIFT belongs to the family of quantum-selected configuration interaction (QSCI) methods. Instead of variationally optimizing a parametrized circuit, the quantum computer is used to sample electronic configurations (Slater determinants) from a set of Krylov states, $e^{-i k \tau H}|\Phi_0\rangle$, prepared with randomized qDRIFT time-evolution circuits. The Hamiltonian is then diagonalized classically in the subspace spanned by the sampled determinants. SHARP (subspace heat-bath adaptive reweighting of probabilities) periodically updates the qDRIFT term-sampling distribution, and optionally the reference state, from the determinants found so far, so that the circuits are steered towards the important parts of the Hilbert space. See the [SqDRIFT explanation](https://qrunch.docs.kvantify.net/docs/explanations/sqdrift.html) in the documentation for the background.

**What you will learn:**
- how to reuse the embedded problems built (and persisted) by the other $\pi$-stacking tutorials;
- how to set up and run SHARP-qDRIFT on a memory-restricted simulator, and what its main knobs do;
- how to compute the binding energy of the dimer, see that it is a correlation effect, and compare it with BEAST-VQE.

This notebook is one of a series of $\pi$-stacking tutorials living in the same folder: the [BEAST-VQE tutorial](pi_stacking_beast_tutorial.ipynb), this SHARP-qDRIFT tutorial, and a quantum phase estimation (QPE) tutorial. They all build exactly the same embedded problems and share the persisted problem files, so the expensive embedding and active space step is only done once, by whichever notebook is run first.

All of the classes and methods we need will be found in the `qrunch` package. The documentation for the Kvantify Qrunch can be found [here](https://qrunch.docs.kvantify.net/index.html). We also define the folders where files will be stored, and set up the logger, which allows us to keep track of basic information when running the algorithms.

```python
import json
from pathlib import Path

import matplotlib.pyplot as plt
import qrunch as qc

result_path = Path("results")
result_path.mkdir(exist_ok=True, parents=True)

output_path = Path("output")
output_path.mkdir(exist_ok=True, parents=True)

qc.setup_logger(qc.LOGGER_INFO)
```

## Hardware requirements
The memory-restricted simulator keeps only a limited number of amplitudes of the state vector, so the memory needed to simulate the circuits is bounded and controlled by a single parameter (see below). The classical part of SHARP-qDRIFT, the repeated diagonalization of the Hamiltonian in the sampled subspace, grows with the number of determinants kept, which is controlled by the pruning options. The embedding and active space step is the most demanding one; its results are stored in the shared `output` folder and reused by all the $\pi$-stacking tutorials.

### Setting up the experiment

The starting point is an .xyz file containing the 17 points of the scan. For the sake of execution time, we only use 7 of them: two points on the repulsive side, the minimum (scan point 2), three points on the attractive side and the dissociated limit (scan point 16). This subselection is the reason for a bit more complicated code in the experiment setup - otherwise we would just pass the xyz file name to `qc.build_reaction_configuration`.

This problem definition is shared across the different pi-stacking tutorials, and running it once will therefore speed up the execution of the other tutorials as well. Building the problem takes about 1h on a standard laptop. The choices are explained in more detail in the [BEAST-VQE tutorial](pi_stacking_beast_tutorial.ipynb); keep them identical across the tutorials, otherwise the persisted problems are not reused.

```python
reaction_path = Path("submodel_dt6dt7_scan.xyz")
# Normally we would use the reaction_path directly in the build_reaction_configuration step, but here we select specific geometries for the 17-point scan; 2 is the minimum, 16 the dissociated limit.
selected_geometries = [0, 1, 2, 4, 6, 9, 16]

def read_xyz_frames(xyz_path: Path) -> list[list[tuple[str, float, float, float]]]:
    """Return the atoms (element, x, y, z) of every frame of a multi-frame .xyz file."""
    lines = xyz_path.read_text().strip().splitlines()
    number_of_atoms = int(lines[0])
    frames = []
    for start in range(0, len(lines), number_of_atoms + 2):
        atom_lines = lines[start + 2 : start + 2 + number_of_atoms]
        frames.append([(element, float(x), float(y), float(z)) for element, x, y, z in map(str.split, atom_lines)])
    return frames

scan_frames = read_xyz_frames(reaction_path)
reaction = [scan_frames[index] for index in selected_geometries]
print(f"Selected scan points {selected_geometries} out of {len(scan_frames)}")
```

Now we build our reaction configuration. Here, we rely on [projective embedding theory](https://qrunch.docs.kvantify.net/docs/explanations/projective_embedding.html): the environment is handled through a DFT mean-field model, while the embedded atoms are treated explicitly with a correlated wave function method, here SHARP-qDRIFT. In order to control this we need to define the embedding space by selecting the atoms that make up the embedded region, and define the spin difference (number of alpha minus beta electrons) and charge states of the system. Finally, we also need to set a basis set for the full calculation, which is modest in size here for acceptable run times of the tutorial. 

The embedded region consists of the two thymine rings, including their carbonyl oxygens and the hydrogens bound to the ring, i.e. the part of the molecules where the $\pi$-stacking happens. The methyl groups and the stubs of the sugar backbone are left to the DFT environment.

```python
bset = "6-31+g"
# Embedded atoms (0-based) of each thymine: ring atoms, carbonyl oxygens and the H atoms on N3 and C6.
ring1 = [2, 3, 4, 5, 6, 7, 8, 10, 24, 25]
ring2 = [12, 13, 14, 15, 16, 17, 18, 20, 27, 28]
embedded_atoms = ring1 + ring2
charge = 0
spin = 0

reaction_config = qc.build_reaction_configuration(
    reaction=reaction,
    basis_set=bset,
    charge=charge,
    spin_difference=spin,
    embedded_atoms=embedded_atoms,
)
```

### Creating the problem with the problem builder

We use the [fluent build pattern](https://qrunch.docs.kvantify.net/docs/explanations/builder_pattern.html) to construct a `problem_builder_creator`. As in the BEAST-VQE tutorial, we calculate along a `reaction_path` with the `even_handed` approach, so that the active orbital assignment is consistent across all frames of the scan. This is crucial for a smooth potential energy surface.

Before going into details here, let us define a couple of parameters that are required for the quantum calculation and which we may want to play with later, specifically:
- the active space: 16 spatial orbitals with 5 alpha (and 5 beta) electrons. Unlike BEAST-VQE, which only describes paired electrons, SHARP-qDRIFT works with the full fermionic wave function, so each spin orbital becomes a qubit and the circuits act on 32 qubits;
- the SHARP-qDRIFT parameters, which are discussed below when the calculator is set up;
- `sharp_geometries`: to keep the runtime low, we run SHARP-qDRIFT on two of the selected geometries only, the minimum of the energy profile and the dissociated limit.

```python
# (16 spatial orbitals, 10 electrons) active space: 32 spin orbitals, i.e. 32 qubits for SHARP-qDRIFT.
active_orbitals = 16
active_alpha_electrons = 5

maximum_krylov_order = 3
number_of_randomizations = 40
subsequence_length = 3
shots_per_circuit = 1000
update_frequency = 20
pruning_threshold = 5e-5
max_number_of_amplitudes = 200_000

# Scan points for SHARP-qDRIFT, a subset of selected_geometries.
sharp_geometries = {2: "minimum", 16: "dissociated dimer"}
```

The choices for the following cell, where we run the fluent builder pattern are:
- a DFT (B3LYP) full-system mean-field calculation;
- an orbital assignment based on the total weight of the atomic orbital contributions;
- MP2 natural orbitals for the embedded region (with a truncation of the virtual space);
- resolution of the identity for the full two-electron repulsion integrals;
- the active space defined above, followed by a conversion to dense integrals (which only happens in the active space).

**Important:** to reuse the orbitals and MP2 calculations that were already performed, every option that determines the stored data (functional, basis set, assignment tolerance, MP2 threshold, auxiliary basis, active space, ...) must be identical to the ones used in the earlier run.

```python
dft_options = qc.options.DftCalculatorOptions(functional="b3lyp")
mp2_options = qc.options.MollerPlesset2CalculatorOptions(virtual_truncation_threshold=1e-6)

persister_path = output_path / "persister_data_sample"
persister_path.mkdir(exist_ok=True, parents=True)

reaction_builder_creator = (
    qc.problem_builder_creator()
    .reaction_path()
    .even_handed()
    .choose_full_system_solver()
    .dft(dft_options)
    .choose_orbital_assigner()
    .total_weight(assignment_tolerance=0.3)
    .choose_embedded_orbital_calculator()
    .moller_plesset_2(mp2_options)
    .choose_repulsion_integral_builder()
    .resolution_of_the_identity(auxiliary_basis="heavy-aug-cc-pvdz-ri")
    .add_problem_modifier()
    .active_space(active_orbitals, active_alpha_electrons)
    .add_problem_modifier()
    .to_dense_integrals()
)
```

When running calculations on embedded systems, it is useful to store the mean-field results and orbital assignment details so that they can be reused. This is the role of the `data_persister_manager`. We point it to the persister folder shared by all the $\pi$-stacking tutorials, and use `load_policy="fallback"`: the stored files are loaded when present, and generated only if they are missing.

```python
reaction_builder_creator = (
    reaction_builder_creator.choose_data_persister_manager().file_persister(
        directory=persister_path,
        extension=".qdk",
        load_policy="fallback",
        overwriting_policy="overwrite",
    )
)

reaction_builder = reaction_builder_creator.create()
```

We now build the even-handed reaction path problem for the selected scan points. If the persisted files are already in place, for instance because one of the other $\pi$-stacking tutorials has been run, this step only loads them (check the log: it should not start new DFT or MP2 calculations). Without existing checkpoint files (the first time this is run) the following cell takes about 1h to finish.

```python
reaction_problem = reaction_builder.build_restricted(reaction_config)
print(f"Built {len(reaction_problem.ground_state_problems)} problems for scan points {selected_geometries}")
```

### Running the quantum algorithm

SHARP-qDRIFT only needs a `sampler`: the energy is obtained by classical diagonalization in the sampled subspace, so no `estimator` is required. We use the `memory_restricted` simulator. Since the qDRIFT time-evolution circuits produce complex amplitudes, we switch off `use_real_numbers`. We keep at most `max_number_of_amplitudes` amplitudes, and prune amplitudes with a norm below `amplitude_cutoff`; tiny amplitudes are essentially never sampled anyway.

```python
sampler_options = qc.options.MemoryRestrictedSimulatorOptions(
    use_real_numbers=False,
    max_number_of_amplitudes=max_number_of_amplitudes,
    amplitude_cutoff=1e-4,
)
sampler = qc.sampler_creator().memory_restricted().with_options(sampler_options).create()
```

Next we set the SqDRIFT options, which define the circuits:
- `maximum_krylov_order`: circuits are generated for every Krylov order $k = 1, \ldots, d$;
- `number_of_randomizations`: the number of independent qDRIFT circuits per Krylov order;
- `subsequence_length`: the number of Hamiltonian terms sampled into each qDRIFT circuit, i.e. its depth in terms of excitation gates.

In total, `maximum_krylov_order * number_of_randomizations` circuits are sampled, each with `shots_per_circuit` shots. A seed makes the randomized circuits reproducible.

```python
sqdrift_options = qc.options.SqDriftOptions(
    subsequence_length=subsequence_length,
    maximum_krylov_order=maximum_krylov_order,
    number_of_randomizations=number_of_randomizations,
    seed=42,
)
```

SHARP-qDRIFT can encode the current subspace ground state in the reference state from which the Krylov circuits start. This improves the accuracy at the cost of deeper circuits. The encoding circuit is found with a beam search over excitation operators; `max_operators` bounds the number of operators, and thus the depth of the state preparation.

```python
state_encoding_options = qc.options.FixedParticleNumberStatePrepBeamSearchOptions(
    beam_width=3,
    max_operators=7,
    max_number_of_amplitudes=max_number_of_amplitudes,
)
```

With those elements, we set up the calculator using a `calculator_creator`. Every `update_frequency` circuits, the sampling strategy is updated: the subspace is diagonalized, the qDRIFT term-sampling distribution is reweighted, and the reference state is re-encoded. At each update, determinants whose configuration-interaction coefficient falls below `pruning_threshold` are removed from the subspace, which keeps the classical diagonalization cheap.

```python
calculator = (
    qc.calculator_creator()
    .quantum_selected_configuration_interaction()
    .sharp_sqdrift()
    .with_sampler(sampler)
    .with_shots_per_circuit(shots_per_circuit)
    .with_sqdrift_options(sqdrift_options)
    .with_update_frequency(update_frequency)
    .with_state_encoding_options(state_encoding_options)
    .with_pruning_options("Coefficient", pruning_threshold=pruning_threshold)
    .create()
)
```

Now the chemistry definitions and the quantum algorithm come together. We run the calculator on the two geometries in `sharp_geometries`. This is the longest step after the problem building, so it is a good moment for a coffee; while you wait, try to guess whether SHARP-qDRIFT will find a deeper or a shallower minimum than BEAST-VQE.

```python
results = {
    scan_point: calculator.calculate(reaction_problem.ground_state_problems[selected_geometries.index(scan_point)])
    for scan_point in sharp_geometries
}
```

### Extracting results

For each geometry, we compare the final total energy with the initial (Hartree-Fock) energy of the active space; the difference is the correlation energy recovered by SHARP-qDRIFT.

```python
sharp_energies = {scan_point: float(result.total_energy.value) for scan_point, result in results.items()}
# The reference (Hartree-Fock) total energy has the same non-active and nuclear contributions as the final one.
reference_energies = {
    scan_point: float(result.initial_active_electrons_energy)
    + float(result.non_active_electrons_energy)
    + float(result.nuclear_repulsion)
    for scan_point, result in results.items()
}

for scan_point, label in sharp_geometries.items():
    correlation = sharp_energies[scan_point] - reference_energies[scan_point]
    print(
        f"Scan point {scan_point:2d} ({label}): E = {sharp_energies[scan_point]:.6f} Ha, "
        f"correlation energy = {correlation * 1000:.2f} mHa"
    )
```

The binding energy of the stacked dimer is the energy difference between the minimum and the dissociated limit. It is a tiny number on the scale of the total energies, which is typical for dispersion-bound systems and makes them a demanding test for any approximate method.

**Where does the binding come from?** We compute the binding energy both with the Hartree-Fock reference and with SHARP-qDRIFT. If the [BEAST-VQE tutorial](pi_stacking_beast_tutorial.ipynb) has been run, its results are loaded from the `results` folder and the two methods are compared on the same geometries. Make a guess before running the cell!

```python
hartree_to_kcal_per_mol = 627.5095
minimum_point, dissociated_point = sharp_geometries

binding_energies = {
    "Reference (Hartree-Fock)": (reference_energies[minimum_point] - reference_energies[dissociated_point])
    * hartree_to_kcal_per_mol,
    "SHARP-qDRIFT": (sharp_energies[minimum_point] - sharp_energies[dissociated_point]) * hartree_to_kcal_per_mol,
}

beast_result_file = result_path / f"pi_stacking_mr_{active_orbitals}o.json"
if beast_result_file.exists():
    with open(beast_result_file) as f:
        beast_data = json.load(f)
    beast_energies = dict(
        zip(beast_data["geometry_indices"], (e["value"] for e in beast_data["total_energy_per_geometry"]))
    )
    if minimum_point in beast_energies and dissociated_point in beast_energies:
        binding_energies["OO-BEAST-VQE + PT2"] = (
            beast_energies[minimum_point] - beast_energies[dissociated_point]
        ) * hartree_to_kcal_per_mol

for method, binding_energy in binding_energies.items():
    print(f"{method:>24s}: binding energy = {binding_energy:.2f} kcal/mol")

plt.figure(figsize=(7, 4.5))
plt.bar(list(binding_energies), list(binding_energies.values()))
plt.axhline(0.0, color="grey", linewidth=0.8)
plt.ylabel("Binding energy (kcal/mol)")
plt.title(f"Scan point {minimum_point} minus scan point {dissociated_point}")
plt.grid(axis="y")
plt.show()
```

As in the BEAST-VQE tutorial, the Hartree-Fock reference shows (essentially) no binding: the stacking of the two thymines only appears once electron correlation is included. How close are the two quantum algorithms to each other? Keep in mind that SHARP-qDRIFT works with all electrons of the active space, while BEAST-VQE describes paired electrons and adds the broken pairs perturbatively.

Finally, we save the results, together with the main parameters of the calculation, to a JSON file.

```python
final_data = {
    "charge": charge,
    "spin": spin,
    "basis_set": str(bset),
    "embedded_atom_indices": embedded_atoms,
    "cas": {
        "number_of_active_spatial_orbitals": active_orbitals,
        "number_of_active_alpha_electrons": active_alpha_electrons,
    },
    "simulator": "memory_restricted",
    "max_number_of_amplitudes": max_number_of_amplitudes,
    "maximum_krylov_order": maximum_krylov_order,
    "number_of_randomizations": number_of_randomizations,
    "subsequence_length": subsequence_length,
    "shots_per_circuit": shots_per_circuit,
    "update_frequency": update_frequency,
    "pruning_threshold": pruning_threshold,
    "geometry_indices": list(sharp_geometries),
    "total_energy_per_geometry": [{"value": sharp_energies[i], "error": 0.0} for i in sharp_geometries],
    "reference_energy_per_geometry": [{"value": reference_energies[i], "error": 0.0} for i in sharp_geometries],
    "nuclear_repulsion_per_geometry": [float(results[i].nuclear_repulsion) for i in sharp_geometries],
    "non_active_electrons_per_geometry": [float(results[i].non_active_electrons_energy) for i in sharp_geometries],
    "binding_energy_kcal_per_mol": binding_energies["SHARP-qDRIFT"],
}

result_name = f"pi_stacking_sharp_sqdrift_{active_orbitals}o.json"
with open(result_path / result_name, "w") as f:
    json.dump(final_data, f, indent=2)
print(f"Saved results to {result_path / result_name}")
```

- **Different scan points:** editing `selected_geometries` changes the problem definition: the problems are built (and persisted) from scratch, which takes about an hour, and they will no longer match the ones used in the other $\pi$-stacking tutorials.
